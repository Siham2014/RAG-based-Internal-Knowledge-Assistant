**1. Fonction de nettoyage**

```python
import re

def clean_text(text: str) -> str:
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()
```

**Étape 1 — Extraire le front matter Markdown**

```python
import re
from typing import Tuple, Dict


def extract_front_matter(raw_text: str) -> Tuple[Dict[str, str], str]:
    """
    Sépare les métadonnées YAML placées au début d'un fichier Markdown
    du contenu documentaire réel.
    """

    metadata = {}

    # Cas standard : métadonnées entourées par ---
    yaml_match = re.match(
        r"^\s*---\s*\n(.*?)\n---\s*\n(.*)$",
        raw_text,
        flags=re.DOTALL
    )

    if yaml_match:
        yaml_block = yaml_match.group(1)
        content = yaml_match.group(2)

    else:
        # Cas des fichiers Microsoft sans séparateur ---
        lines = raw_text.splitlines()
        yaml_lines = []
        content_start = 0

        metadata_pattern = re.compile(
            r"^[A-Za-z0-9_.-]+\s*:"
        )

        for index, line in enumerate(lines):
            stripped = line.strip()

            if not stripped:
                if yaml_lines:
                    content_start = index + 1
                    break
                continue

            if metadata_pattern.match(stripped) or stripped.startswith("- "):
                yaml_lines.append(line)
            else:
                content_start = index
                break

        yaml_block = "\n".join(yaml_lines)
        content = "\n".join(lines[content_start:])

    # Extraction simple des champs clé: valeur
    for line in yaml_block.splitlines():
        match = re.match(r"^([A-Za-z0-9_.-]+)\s*:\s*(.*)$", line.strip())

        if match:
            key = match.group(1).strip()
            value = match.group(2).strip().strip("\"'")

            if value:
                metadata[key] = value

    return metadata, content
```

**2. Configuration expérimentale**

```python
CHUNK_SIZES = [256, 512, 1024]
OVERLAP_RATIO = 0.20

print("Tailles testées :", CHUNK_SIZES)
print("Overlap :", int(OVERLAP_RATIO * 100), "%")
```

**3. Fonction pour créer les métadonnées des chunks**

```python
def build_chunk(
    text: str,
    original_doc: Document,
    chunk_index: int,
    strategy: str,
    chunk_size: int,
    chunk_overlap: int
) -> Document:
    """
    Crée un chunk LangChain avec des métadonnées de traçabilité.
    """

    metadata = original_doc.metadata.copy()

    source_name = metadata.get("source", "unknown_source")
    page = metadata.get("page")

    if page is not None:
        chunk_id = (
            f"{strategy}__{chunk_size}__"
            f"{source_name}__page_{page}__chunk_{chunk_index}"
        )
    else:
        chunk_id = (
            f"{strategy}__{chunk_size}__"
            f"{source_name}__chunk_{chunk_index}"
        )

    metadata.update({
        "chunk_id": chunk_id,
        "chunk_index": chunk_index,
        "strategy": strategy,
        "configured_chunk_size": chunk_size,
        "configured_overlap": chunk_overlap,
        "actual_characters": len(text),
        "actual_words": len(text.split())
    })

    return Document(
        page_content=text,
        metadata=metadata
    )
```

**4. Implémenter le Fixed-size Chunking**

```python
def fixed_size_chunking(
    documents: list[Document],
    chunk_size: int,
    chunk_overlap: int
) -> list[Document]:
    """
    Découpe chaque document avec une fenêtre fixe de caractères.
    """

    if chunk_size <= 0:
        raise ValueError("chunk_size doit être supérieur à 0.")

    if chunk_overlap < 0:
        raise ValueError("chunk_overlap ne peut pas être négatif.")

    if chunk_overlap >= chunk_size:
        raise ValueError(
            "chunk_overlap doit être strictement inférieur à chunk_size."
        )

    step = chunk_size - chunk_overlap
    chunks = []

    for doc in documents:
        text = doc.page_content.strip()

        if not text:
            continue

        chunk_index = 0

        for start in range(0, len(text), step):
            end = start + chunk_size
            chunk_text = text[start:end].strip()

            if not chunk_text:
                continue

            chunks.append(
                build_chunk(
                    text=chunk_text,
                    original_doc=doc,
                    chunk_index=chunk_index,
                    strategy="fixed",
                    chunk_size=chunk_size,
                    chunk_overlap=chunk_overlap
                )
            )

            chunk_index += 1

            if end >= len(text):
                break

    return chunks
```

**5. Implémenter le Recursive Chunking**

```python
def recursive_chunking(
    documents: list[Document],
    chunk_size: int,
    chunk_overlap: int
) -> list[Document]:
    """
    Découpe les documents en privilégiant les frontières naturelles.
    """

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        separators=[
            "\n\n",
            "\n",
            ". ",
            "? ",
            "! ",
            "; ",
            ", ",
            " ",
            ""
        ],
        length_function=len,
        keep_separator=True,
        is_separator_regex=False
    )

    chunks = []

    for doc in documents:
        text = doc.page_content.strip()

        if not text:
            continue

        split_texts = splitter.split_text(text)

        for chunk_index, chunk_text in enumerate(split_texts):
            chunk_text = chunk_text.strip()

            if not chunk_text:
                continue

            chunks.append(
                build_chunk(
                    text=chunk_text,
                    original_doc=doc,
                    chunk_index=chunk_index,
                    strategy="recursive",
                    chunk_size=chunk_size,
                    chunk_overlap=chunk_overlap
                )
            )

    return chunks
```

**6. Exécuter les expériences Fixed et Recursive**

```python
chunking_results = {}
execution_records = []

for chunk_size in CHUNK_SIZES:
    chunk_overlap = int(chunk_size * OVERLAP_RATIO)

    # Fixed-size
    start_time = time.perf_counter()

    fixed_chunks = fixed_size_chunking(
        documents=documents,
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap
    )

    fixed_time = time.perf_counter() - start_time

    chunking_results[("fixed", chunk_size)] = fixed_chunks

    execution_records.append({
        "strategy": "fixed",
        "chunk_size": chunk_size,
        "chunk_overlap": chunk_overlap,
        "number_of_chunks": len(fixed_chunks),
        "chunking_time_seconds": fixed_time
    })

    # Recursive
    start_time = time.perf_counter()

    recursive_chunks = recursive_chunking(
        documents=documents,
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap
    )

    recursive_time = time.perf_counter() - start_time

    chunking_results[("recursive", chunk_size)] = recursive_chunks

    execution_records.append({
        "strategy": "recursive",
        "chunk_size": chunk_size,
        "chunk_overlap": chunk_overlap,
        "number_of_chunks": len(recursive_chunks),
        "chunking_time_seconds": recursive_time
    })

print("Expériences Fixed et Recursive terminées ")
```

**8. Calculer les statistiques détaillées**

```python
def calculate_chunk_statistics(
    chunks: list[Document],
    strategy: str,
    chunk_size: int
) -> dict:
    """
    Calcule les statistiques descriptives d'une collection de chunks.
    """

    character_lengths = np.array([
        len(chunk.page_content)
        for chunk in chunks
    ])

    word_lengths = np.array([
        len(chunk.page_content.split())
        for chunk in chunks
    ])

    return {
        "strategy": strategy,
        "chunk_size": chunk_size,
        "number_of_chunks": len(chunks),

        "mean_characters": round(character_lengths.mean(), 2),
        "median_characters": round(float(np.median(character_lengths)), 2),
        "min_characters": int(character_lengths.min()),
        "max_characters": int(character_lengths.max()),
        "std_characters": round(character_lengths.std(), 2),

        "mean_words": round(word_lengths.mean(), 2),
        "median_words": round(float(np.median(word_lengths)), 2),
        "min_words": int(word_lengths.min()),
        "max_words": int(word_lengths.max())
    }
```

```python
statistics_records = []

for (strategy, chunk_size), chunks in chunking_results.items():
    statistics_records.append(
        calculate_chunk_statistics(
            chunks=chunks,
            strategy=strategy,
            chunk_size=chunk_size
        )
    )

statistics_df = pd.DataFrame(statistics_records)

statistics_df = statistics_df.sort_values(
    by=["chunk_size", "strategy"]
).reset_index(drop=True)

statistics_df
```

```python
#verefication automatique
def validate_chunks(
    chunks: list[Document],
    expected_strategy: str,
    expected_chunk_size: int
) -> dict:

    empty_chunks = [
        chunk for chunk in chunks
        if not chunk.page_content.strip()
    ]

    duplicated_ids = (
        len({
            chunk.metadata["chunk_id"]
            for chunk in chunks
        }) != len(chunks)
    )

    oversized_chunks = [
        chunk for chunk in chunks
        if len(chunk.page_content) > expected_chunk_size
    ]

    incorrect_strategies = [
        chunk for chunk in chunks
        if chunk.metadata["strategy"] != expected_strategy
    ]

    return {
        "strategy": expected_strategy,
        "chunk_size": expected_chunk_size,
        "total_chunks": len(chunks),
        "empty_chunks": len(empty_chunks),
        "duplicated_chunk_ids": duplicated_ids,
        "oversized_chunks": len(oversized_chunks),
        "incorrect_strategy_metadata": len(incorrect_strategies)
    }
```

```python
validation_records = []

for (strategy, chunk_size), chunks in chunking_results.items():
    validation_records.append(
        validate_chunks(
            chunks=chunks,
            expected_strategy=strategy,
            expected_chunk_size=chunk_size
        )
    )

validation_df = pd.DataFrame(validation_records)

validation_df
```

**Fonction de Semantic Chunking**

```python
def semantic_chunking(
    documents: list[Document],
    embeddings,
    breakpoint_threshold_amount: float
) -> list[Document]:
    """
    Découpe les documents selon les changements sémantiques
    entre groupes de phrases.
    """

    splitter = SemanticChunker(
        embeddings=embeddings,
        breakpoint_threshold_type="percentile",
        breakpoint_threshold_amount=breakpoint_threshold_amount,
        sentence_split_regex=r"(?<=[.?!])\s+"
    )

    semantic_chunks = []

    for doc in documents:
        text = doc.page_content.strip()

        if not text:
            continue

        try:
            split_documents = splitter.create_documents(
                texts=[text],
                metadatas=[doc.metadata.copy()]
            )

        except Exception as error:
            print(
                "Erreur pour le document :",
                doc.metadata.get("source")
            )
            print(type(error).__name__, ":", error)
            continue

        for chunk_index, split_doc in enumerate(split_documents):
            chunk_text = split_doc.page_content.strip()

            if not chunk_text:
                continue

            metadata = split_doc.metadata.copy()

            source_name = metadata.get(
                "source",
                "unknown_source"
            )

            page = metadata.get("page")

            if page is not None:
                chunk_id = (
                    f"semantic__p{breakpoint_threshold_amount}__"
                    f"{source_name}__page_{page}__"
                    f"chunk_{chunk_index}"
                )
            else:
                chunk_id = (
                    f"semantic__p{breakpoint_threshold_amount}__"
                    f"{source_name}__chunk_{chunk_index}"
                )

            metadata.update({
                "chunk_id": chunk_id,
                "chunk_index": chunk_index,
                "strategy": "semantic",
                "semantic_threshold_type": "percentile",
                "semantic_threshold_amount":
                    breakpoint_threshold_amount,
                "configured_chunk_size": None,
                "configured_overlap": 0,
                "actual_characters": len(chunk_text),
                "actual_words": len(chunk_text.split())
            })

            semantic_chunks.append(
                Document(
                    page_content=chunk_text,
                    metadata=metadata
                )
            )

    return semantic_chunks
```

**Calculer les statistiques des chunks sémantiques**

```python
def calculate_semantic_statistics(
    chunks: list[Document],
    percentile: int
) -> dict:

    character_lengths = np.array([
        len(chunk.page_content)
        for chunk in chunks
    ])

    word_lengths = np.array([
        len(chunk.page_content.split())
        for chunk in chunks
    ])

    return {
        "strategy": "semantic",
        "configuration": f"percentile_{percentile}",
        "semantic_percentile": percentile,
        "number_of_chunks": len(chunks),

        "mean_characters":
            round(float(character_lengths.mean()), 2),

        "median_characters":
            round(float(np.median(character_lengths)), 2),

        "min_characters":
            int(character_lengths.min()),

        "max_characters":
            int(character_lengths.max()),

        "std_characters":
            round(float(character_lengths.std()), 2),

        "mean_words":
            round(float(word_lengths.mean()), 2),

        "median_words":
            round(float(np.median(word_lengths)), 2),

        "min_words":
            int(word_lengths.min()),

        "max_words":
            int(word_lengths.max())
    }
```

```python
semantic_statistics_records = []

for percentile in SEMANTIC_PERCENTILES:
    chunks = chunking_results[
        ("semantic", percentile)
    ]

    semantic_statistics_records.append(
        calculate_semantic_statistics(
            chunks=chunks,
            percentile=percentile
        )
    )

semantic_statistics_df = pd.DataFrame(
    semantic_statistics_records
)

semantic_statistics_df
```

**Valider les chunks sémantiques**

```python
def validate_semantic_chunks(
    chunks: list[Document],
    percentile: int
) -> dict:

    empty_chunks = [
        chunk
        for chunk in chunks
        if not chunk.page_content.strip()
    ]

    chunk_ids = [
        chunk.metadata["chunk_id"]
        for chunk in chunks
    ]

    duplicated_ids = (
        len(chunk_ids) != len(set(chunk_ids))
    )

    incorrect_strategy = [
        chunk
        for chunk in chunks
        if chunk.metadata.get("strategy") != "semantic"
    ]

    very_small_chunks = [
        chunk
        for chunk in chunks
        if len(chunk.page_content) < 50
    ]

    return {
        "strategy": "semantic",
        "semantic_percentile": percentile,
        "total_chunks": len(chunks),
        "empty_chunks": len(empty_chunks),
        "duplicated_chunk_ids": duplicated_ids,
        "incorrect_strategy_metadata":
            len(incorrect_strategy),
        "chunks_under_50_characters":
            len(very_small_chunks)
    }
```

```python
semantic_validation_records = []

for percentile in SEMANTIC_PERCENTILES:
    chunks = chunking_results[
        ("semantic", percentile)
    ]

    semantic_validation_records.append(
        validate_semantic_chunks(
            chunks=chunks,
            percentile=percentile
        )
    )

semantic_validation_df = pd.DataFrame(
    semantic_validation_records
)

semantic_validation_df
```

**Étape 2 — Fonction pour explorer un document:**
****
Cette fonction permet d’afficher le contenu d’un fichier afin de créer des questions fondées sur son contenu réel.

```python
def inspect_source(
    source_keyword: str,
    max_characters: int = 5000,
    start_unit: int = 0,
    max_units: int = 3
) -> None:
    """
    Affiche une partie des unités documentaires correspondant
    à une source.

    Utile notamment pour les PDF composés de nombreuses pages.
    """

    matching_documents = [
        doc
        for doc in documents
        if source_keyword.lower()
        in str(
            doc.metadata.get("source", "")
        ).lower()
    ]

    print(
        "Unités documentaires trouvées :",
        len(matching_documents)
    )

    selected_documents = matching_documents[
        start_unit:start_unit + max_units
    ]

    if not selected_documents:
        print("Aucune unité dans cet intervalle.")
        return

    for index, doc in enumerate(
        selected_documents,
        start=start_unit + 1
    ):
        print("\n" + "=" * 100)
        print("Unité :", index)
        print("Source :", doc.metadata.get("source"))
        print("Format :", doc.metadata.get("format"))
        print("Titre :", doc.metadata.get("title", ""))
        print("Page :", doc.metadata.get("page", "N/A"))
        print("-" * 100)
        print(doc.page_content[:max_characters])
```

```python
import re
import unicodedata


def normalize_for_matching(text: str) -> str:
    """
    Normalise le texte pour ignorer :
    - les majuscules/minuscules ;
    - les espaces multiples ;
    - les retours à la ligne ;
    - les variantes Unicode ;
    - les différents types de tirets.
    """

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    ).lower()

    # Uniformiser les tirets Unicode
    text = re.sub(
        r"[‐-‒–—−]",
        "-",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


def verify_gold_evidence(row) -> bool:
    """
    Vérifie que le passage de référence existe réellement
    dans le document attendu.
    """

    expected_source = row[
        "expected_source_canonical"
    ]

    matching_documents = [
        doc
        for doc in documents
        if canonical_source(
            doc.metadata.get("source", "")
        ) == expected_source
    ]

    normalized_evidence = normalize_for_matching(
        row["gold_evidence"]
    )

    if not normalized_evidence:
        return False

    for doc in matching_documents:

        normalized_content = normalize_for_matching(
            doc.page_content
        )

        if normalized_evidence in normalized_content:
            return True

    return False


evaluation_df["evidence_exists"] = (
    evaluation_df.apply(
        verify_gold_evidence,
        axis=1
    )
)


evidence_validation_df = evaluation_df[
    [
        "id",
        "format",
        "source_exists",
        "evidence_exists"
    ]
]


display(evidence_validation_df)


invalid_questions_df = evaluation_df[
    (
        evaluation_df["source_exists"] == False
    )
    |
    (
        evaluation_df["evidence_exists"] == False
    )
]


print(
    "Questions invalides :",
    len(invalid_questions_df)
)

if not invalid_questions_df.empty:
    display(
        invalid_questions_df[
            [
                "id",
                "format",
                "expected_source",
                "source_exists",
                "evidence_exists",
                "gold_evidence"
            ]
        ]
    )
```

**Étape 13 — Définir la pertinence d’un résultat**

*Pertinence au niveau de la source*

```python
def is_source_relevant(
    result: dict,
    expected_source: str
) -> bool:
    """
    Vérifie si le résultat vient du document attendu,
    indépendamment du chemin Windows ou Linux.
    """

    retrieved_source = canonical_source(
        result.get("source", "")
    )

    expected_source_normalized = canonical_source(
        expected_source
    )

    return (
        retrieved_source
        == expected_source_normalized
    )
```

**Pertinence au niveau du passage de preuve**

```python
import re
import unicodedata


def normalize_evaluation_text(
    text: str
) -> str:

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    ).lower()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()
```

```python
def evidence_token_recall(
    retrieved_text: str,
    gold_evidence: str
) -> float:
    """
    Proportion des mots distinctifs du passage de référence
    présents dans le chunk récupéré.
    """

    retrieved_tokens = set(
        normalize_evaluation_text(
            retrieved_text
        ).split()
    )

    evidence_tokens = set(
        normalize_evaluation_text(
            gold_evidence
        ).split()
    )

    if not evidence_tokens:
        return 0.0

    common_tokens = (
        retrieved_tokens
        & evidence_tokens
    )

    return (
        len(common_tokens)
        / len(evidence_tokens)
    )
```

```python
EVIDENCE_THRESHOLD = 0.60


def is_evidence_relevant(
    result: dict,
    expected_source: str,
    gold_evidence: str,
    threshold: float = EVIDENCE_THRESHOLD
) -> bool:
    """
    Vérifie simultanément :
    1. que le résultat vient de la bonne source ;
    2. qu'il contient suffisamment du passage attendu.
    """

    if not is_source_relevant(
        result=result,
        expected_source=expected_source
    ):
        return False

    normalized_chunk = normalize_for_matching(
        result.get("text", "")
    )

    normalized_evidence = normalize_for_matching(
        gold_evidence
    )

    if not normalized_evidence:
        return False

    # Cas idéal : passage complet présent.
    if normalized_evidence in normalized_chunk:
        return True

    # Cas où une frontière de chunk coupe le passage.
    overlap_score = evidence_token_recall(
        retrieved_text=result.get(
            "text",
            ""
        ),
        gold_evidence=gold_evidence
    )

    return overlap_score >= threshold
```

**Étape 14 — Calculer Recall@K et MRR**

```python
def reciprocal_rank(
    relevance_values: list[bool]
) -> float:

    for rank, relevant in enumerate(
        relevance_values,
        start=1
    ):
        if relevant:
            return 1.0 / rank

    return 0.0
```

```python
import faiss
import numpy as np


def search_faiss(
    question: str,
    index,
    chunks,
    embeddings,
    top_k: int = 5
) -> list[dict]:
    """
    Encode une question, recherche les chunks les plus proches
    dans l'index FAISS et retourne les résultats classés.
    """

    if not question or not question.strip():
        raise ValueError(
            "La question ne peut pas être vide."
        )

    if top_k <= 0:
        raise ValueError(
            "top_k doit être strictement positif."
        )

    if index.ntotal != len(chunks):
        raise ValueError(
            "Le nombre de vecteurs FAISS ne correspond pas "
            "au nombre de chunks : "
            f"{index.ntotal} vecteurs contre {len(chunks)} chunks."
        )

    # Créer l'embedding de la question
    query_vector = embeddings.embed_query(
        question
    )

    query_vector = np.asarray(
        [query_vector],
        dtype=np.float32
    )

    if query_vector.ndim != 2:
        raise ValueError(
            "La forme du vecteur de requête est invalide."
        )

    # Sécurité : normalisation L2 pour IndexFlatIP
    faiss.normalize_L2(
        query_vector
    )

    effective_top_k = min(
        top_k,
        index.ntotal
    )

    scores, indices = index.search(
        query_vector,
        effective_top_k
    )

    results = []

    for rank, (
        chunk_index,
        score
    ) in enumerate(
        zip(indices[0], scores[0]),
        start=1
    ):
        if chunk_index < 0:
            continue

        chunk = chunks[
            int(chunk_index)
        ]

        results.append({
            "rank": rank,
            "score": float(score),
            "chunk_index": int(chunk_index),
            "chunk_id": chunk.metadata.get(
                "chunk_id"
            ),
            "source": chunk.metadata.get(
                "source"
            ),
            "text": chunk.page_content
        })

    return results


print(
    "✅ Fonction search_faiss définie"
)
```

```python
def evaluate_single_configuration(
    evaluation_data: pd.DataFrame,
    index,
    chunks: list[Document],
    embeddings,
    configuration_name: str,
    retrieval_k: int = 10,
    evidence_threshold: float = 0.60
):
    """
    Évalue une configuration avec :
    - Source Recall@1, @3, @5, @10
    - Evidence Recall@1, @3, @5, @10
    - Source MRR
    - Evidence MRR
    - latence moyenne
    """

    question_records = []

    for _, row in evaluation_data.iterrows():
        start_time = time.perf_counter()

        results = search_faiss(
            question=row["question"],
            index=index,
            chunks=chunks,
            embeddings=embeddings,
            top_k=retrieval_k
        )

        retrieval_time = (
            time.perf_counter() - start_time
        )

        source_relevance = [
            is_source_relevant(
                result=result,
                expected_source=row["expected_source"]
            )
            for result in results
        ]

        evidence_relevance = [
            is_evidence_relevant(
                result=result,
                expected_source=row["expected_source"],
                gold_evidence=row["gold_evidence"],
                threshold=evidence_threshold
            )
            for result in results
        ]

        record = {
            "configuration": configuration_name,
            "question_id": row["id"],
            "question": row["question"],
            "expected_source": row["expected_source"],
            "retrieval_time_seconds": retrieval_time,

            "source_recall_at_1":
                int(any(source_relevance[:1])),

            "source_recall_at_3":
                int(any(source_relevance[:3])),

            "source_recall_at_5":
                int(any(source_relevance[:5])),

            "source_recall_at_10":
                int(any(source_relevance[:10])),

            "evidence_recall_at_1":
                int(any(evidence_relevance[:1])),

            "evidence_recall_at_3":
                int(any(evidence_relevance[:3])),

            "evidence_recall_at_5":
                int(any(evidence_relevance[:5])),

            "evidence_recall_at_10":
                int(any(evidence_relevance[:10])),

            "source_reciprocal_rank":
                reciprocal_rank(source_relevance),

            "evidence_reciprocal_rank":
                reciprocal_rank(evidence_relevance),

            "top_1_source":
                results[0]["source"]
                if results else None,

            "top_1_score":
                results[0]["score"]
                if results else None
        }

        question_records.append(record)

    detail_df = pd.DataFrame(
        question_records
    )

    summary = {
        "configuration": configuration_name,
        "number_of_questions": len(detail_df),

        "source_recall_at_1":
            detail_df["source_recall_at_1"].mean(),

        "source_recall_at_3":
            detail_df["source_recall_at_3"].mean(),

        "source_recall_at_5":
            detail_df["source_recall_at_5"].mean(),

        "source_recall_at_10":
            detail_df["source_recall_at_10"].mean(),

        "evidence_recall_at_1":
            detail_df["evidence_recall_at_1"].mean(),

        "evidence_recall_at_3":
            detail_df["evidence_recall_at_3"].mean(),

        "evidence_recall_at_5":
            detail_df["evidence_recall_at_5"].mean(),

        "evidence_recall_at_10":
            detail_df["evidence_recall_at_10"].mean(),

        "source_mrr":
            detail_df[
                "source_reciprocal_rank"
            ].mean(),

        "evidence_mrr":
            detail_df[
                "evidence_reciprocal_rank"
            ].mean(),

        "mean_retrieval_time_ms":
            detail_df[
                "retrieval_time_seconds"
            ].mean() * 1000,

        "p95_retrieval_time_ms":
            detail_df[
                "retrieval_time_seconds"
            ].quantile(0.95) * 1000
    }

    summary_df = pd.DataFrame(
        [summary]
    )

    return summary_df, detail_df
```

```python
import gc
import json
import time
from pathlib import Path

import faiss
import numpy as np
import pandas as pd


# ============================================================
# 1. Dossiers
# ============================================================

EMBEDDING_CACHE_DIR = Path(
    "/content/embedding_cache"
)

EMBEDDING_CACHE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


RESULTS_DIR = Path(
    "/content/chunking_benchmark_results"
)

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. Paramètres
# ============================================================

TOP_K = 10
EVIDENCE_THRESHOLD = 0.60


# ============================================================
# 3. Contrôles avant exécution
# ============================================================

assert len(evaluation_df) == 30, (
    f"Le benchmark nécessite 30 questions, "
    f"mais {len(evaluation_df)} ont été trouvées."
)

assert evaluation_df["source_exists"].all(), (
    "Certaines sources sont invalides."
)

assert evaluation_df["evidence_exists"].all(), (
    "Certaines preuves sont invalides."
)

assert len(chunking_results) == 9, (
    f"Le benchmark nécessite 9 configurations, "
    f"mais {len(chunking_results)} ont été trouvées."
)

assert "search_faiss" in globals(), (
    "La fonction search_faiss n'est pas définie."
)

assert "evaluate_single_configuration" in globals(), (
    "La fonction evaluate_single_configuration "
    "n'est pas définie."
)

assert "EMBEDDING_MODEL_NAME" in globals(), (
    "La variable EMBEDDING_MODEL_NAME "
    "n'est pas définie."
)


# ============================================================
# 4. Préparation
# ============================================================

all_summary_results = []
all_detail_results = []

benchmark_start_time = time.perf_counter()


configuration_items = sorted(
    chunking_results.items(),
    key=lambda item: (
        str(item[0][0]),
        int(item[0][1])
    )
)


number_of_configurations = len(
    configuration_items
)


print(
    "Configurations à tester :",
    number_of_configurations
)

print(
    "Questions par configuration :",
    len(evaluation_df)
)

print(
    "Nombre total de recherches :",
    number_of_configurations
    * len(evaluation_df)
)


# ============================================================
# 5. Boucle des 9 configurations
# ============================================================

for configuration_number, (
    configuration_key,
    chunks
) in enumerate(
    configuration_items,
    start=1
):

    strategy = str(
        configuration_key[0]
    )

    parameter = int(
        configuration_key[1]
    )

    configuration_name = (
        f"{strategy}_{parameter}"
    )


    print("\n" + "=" * 100)

    print(
        f"Configuration "
        f"{configuration_number}/"
        f"{number_of_configurations} : "
        f"{configuration_name}"
    )

    print(
        "Nombre de chunks :",
        len(chunks)
    )

    print("=" * 100)


    configuration_start_time = (
        time.perf_counter()
    )


    # --------------------------------------------------------
    # A. Préparer les textes
    # --------------------------------------------------------

    chunk_texts = [
        chunk.page_content.strip()
        for chunk in chunks
    ]


    empty_chunk_count = sum(
        1
        for text in chunk_texts
        if not text
    )


    if empty_chunk_count > 0:
        raise ValueError(
            f"{configuration_name} contient "
            f"{empty_chunk_count} chunks vides."
        )


    print("Chunks vides : 0")


    # --------------------------------------------------------
    # B. Construire les chemins du cache
    # --------------------------------------------------------

    safe_model_name = (
        EMBEDDING_MODEL_NAME
        .replace("/", "__")
        .replace("\\", "__")
    )


    embedding_cache_path = (
        EMBEDDING_CACHE_DIR
        / (
            f"{configuration_name}"
            f"__{safe_model_name}.npy"
        )
    )


    metadata_cache_path = (
        EMBEDDING_CACHE_DIR
        / (
            f"{configuration_name}"
            f"__{safe_model_name}.json"
        )
    )


    embedding_loaded_from_cache = False


    # --------------------------------------------------------
    # C. Charger ou générer les embeddings
    # --------------------------------------------------------

    embedding_start_time = (
        time.perf_counter()
    )


    if embedding_cache_path.exists():

        print(
            "Cache trouvé :",
            embedding_cache_path.name
        )


        chunk_vectors = np.load(
            embedding_cache_path,
            allow_pickle=False
        ).astype(
            np.float32,
            copy=False
        )


        embedding_loaded_from_cache = True


        embedding_time_seconds = (
            time.perf_counter()
            - embedding_start_time
        )


    else:

        print(
            "Cache absent : calcul des embeddings"
        )


        chunk_vectors = (
            embedding_model.embed_documents(
                chunk_texts
            )
        )


        chunk_vectors = np.asarray(
            chunk_vectors,
            dtype=np.float32
        )


        if chunk_vectors.ndim != 2:
            raise ValueError(
                f"Embeddings invalides pour "
                f"{configuration_name} : "
                f"{chunk_vectors.shape}"
            )


        # Normalisation avant sauvegarde
        faiss.normalize_L2(
            chunk_vectors
        )


        embedding_time_seconds = (
            time.perf_counter()
            - embedding_start_time
        )


        np.save(
            embedding_cache_path,
            chunk_vectors
        )


        cache_metadata = {
            "configuration": configuration_name,
            "strategy": strategy,
            "parameter": parameter,
            "model_name": EMBEDDING_MODEL_NAME,
            "number_of_chunks": len(chunks),
            "embedding_dimension": int(
                chunk_vectors.shape[1]
            ),
            "normalized": True,
            "embedding_time_seconds": float(
                embedding_time_seconds
            )
        }


        with open(
            metadata_cache_path,
            "w",
            encoding="utf-8"
        ) as metadata_file:

            json.dump(
                cache_metadata,
                metadata_file,
                indent=4,
                ensure_ascii=False
            )


        print(
            "Cache enregistré :",
            embedding_cache_path.name
        )


    # --------------------------------------------------------
    # D. Vérifier le cache ou les embeddings calculés
    # --------------------------------------------------------

    if chunk_vectors.ndim != 2:
        raise ValueError(
            f"Embeddings invalides pour "
            f"{configuration_name} : "
            f"{chunk_vectors.shape}"
        )


    if chunk_vectors.shape[0] != len(chunks):
        raise ValueError(
            f"Cache incompatible pour "
            f"{configuration_name} : "
            f"{chunk_vectors.shape[0]} vecteurs "
            f"pour {len(chunks)} chunks."
        )


    embedding_dimension = int(
        chunk_vectors.shape[1]
    )


    print(
        "Forme des embeddings :",
        chunk_vectors.shape
    )

    print(
        "Chargés depuis le cache :",
        embedding_loaded_from_cache
    )

    print(
        "Temps chargement/calcul :",
        round(
            embedding_time_seconds,
            3
        ),
        "secondes"
    )


    # --------------------------------------------------------
    # E. Vérifier puis normaliser les vecteurs
    # --------------------------------------------------------

    vector_norms_before = np.linalg.norm(
        chunk_vectors,
        axis=1
    )


    mean_vector_norm = float(
        vector_norms_before.mean()
    )


    print(
        "Norme moyenne avant FAISS :",
        round(
            mean_vector_norm,
            4
        )
    )


    # Sécurité supplémentaire
    faiss.normalize_L2(
        chunk_vectors
    )


    # --------------------------------------------------------
    # F. Construire l'index FAISS
    # --------------------------------------------------------

    faiss_index = faiss.IndexFlatIP(
        embedding_dimension
    )


    indexing_start_time = (
        time.perf_counter()
    )


    faiss_index.add(
        chunk_vectors
    )


    faiss_indexing_time_seconds = (
        time.perf_counter()
        - indexing_start_time
    )


    print(
        "Vecteurs indexés :",
        faiss_index.ntotal
    )

    print(
        "Temps d'indexation FAISS :",
        round(
            faiss_indexing_time_seconds,
            4
        ),
        "secondes"
    )


    # --------------------------------------------------------
    # G. Évaluer les 30 questions
    # --------------------------------------------------------

    summary_df, details_df = (
        evaluate_single_configuration(
            evaluation_data=evaluation_df,
            index=faiss_index,
            chunks=chunks,
            embeddings=embedding_model,
            configuration_name=configuration_name,
            retrieval_k=TOP_K,
            evidence_threshold=EVIDENCE_THRESHOLD
        )
    )


    total_configuration_time_seconds = (
        time.perf_counter()
        - configuration_start_time
    )


    # --------------------------------------------------------
    # H. Ajouter les informations techniques
    # --------------------------------------------------------

    summary_df["strategy"] = strategy

    summary_df["parameter"] = parameter


    summary_df[
        "number_of_chunks"
    ] = len(chunks)


    summary_df[
        "embedding_model"
    ] = EMBEDDING_MODEL_NAME


    summary_df[
        "embedding_dimension"
    ] = embedding_dimension


    summary_df[
        "embedding_time_seconds"
    ] = embedding_time_seconds


    summary_df[
        "embedding_loaded_from_cache"
    ] = embedding_loaded_from_cache


    summary_df[
        "embedding_cache_path"
    ] = str(
        embedding_cache_path
    )


    summary_df[
        "metadata_cache_path"
    ] = str(
        metadata_cache_path
    )


    summary_df[
        "faiss_indexing_time_seconds"
    ] = faiss_indexing_time_seconds


    summary_df[
        "total_configuration_time_seconds"
    ] = total_configuration_time_seconds


    summary_df[
        "mean_vector_norm"
    ] = mean_vector_norm


    summary_df[
        "questions_markdown"
    ] = int(
        (
            evaluation_df["format"]
            == "markdown"
        ).sum()
    )


    summary_df[
        "questions_html"
    ] = int(
        (
            evaluation_df["format"]
            == "html"
        ).sum()
    )


    summary_df[
        "questions_pdf"
    ] = int(
        (
            evaluation_df["format"]
            == "pdf"
        ).sum()
    )


    details_df["format"] = (
        details_df["expected_source"]
        .apply(detect_format)
    )


    details_df[
        "embedding_model"
    ] = EMBEDDING_MODEL_NAME


    details_df[
        "embedding_loaded_from_cache"
    ] = embedding_loaded_from_cache


    # --------------------------------------------------------
    # I. Stocker les résultats
    # --------------------------------------------------------

    all_summary_results.append(
        summary_df.copy()
    )


    all_detail_results.append(
        details_df.copy()
    )


    print("\nRésultats :")


    print(
        "Evidence Recall@1 :",
        round(
            summary_df[
                "evidence_recall_at_1"
            ].iloc[0],
            4
        )
    )


    print(
        "Evidence Recall@3 :",
        round(
            summary_df[
                "evidence_recall_at_3"
            ].iloc[0],
            4
        )
    )


    print(
        "Evidence Recall@5 :",
        round(
            summary_df[
                "evidence_recall_at_5"
            ].iloc[0],
            4
        )
    )


    print(
        "Evidence MRR :",
        round(
            summary_df[
                "evidence_mrr"
            ].iloc[0],
            4
        )
    )


    print(
        "Temps total :",
        round(
            total_configuration_time_seconds,
            2
        ),
        "secondes"
    )


    # --------------------------------------------------------
    # J. Sauvegarde intermédiaire
    # --------------------------------------------------------

    progress_summary_df = pd.concat(
        all_summary_results,
        ignore_index=True
    )


    progress_details_df = pd.concat(
        all_detail_results,
        ignore_index=True
    )


    progress_summary_df.to_csv(
        RESULTS_DIR
        / "chunking_summary_progress.csv",
        index=False,
        encoding="utf-8"
    )


    progress_details_df.to_csv(
        RESULTS_DIR
        / "chunking_details_progress.csv",
        index=False,
        encoding="utf-8"
    )


    print(
        "Progression sauvegardée :",
        configuration_name
    )


    # --------------------------------------------------------
    # K. Libérer la mémoire
    # --------------------------------------------------------

    del chunk_texts
    del chunk_vectors
    del faiss_index
    del summary_df
    del details_df
    del progress_summary_df
    del progress_details_df


    gc.collect()


# ============================================================
# 6. Fin du benchmark
# ============================================================

benchmark_total_time_seconds = (
    time.perf_counter()
    - benchmark_start_time
)


print("\n" + "=" * 100)

print(
    "✅ BENCHMARK DES 9 CONFIGURATIONS TERMINÉ"
)

print(
    "Temps total :",
    round(
        benchmark_total_time_seconds / 60,
        2
    ),
    "minutes"
)

print("=" * 100)
```

**Ajouter les résultats par format**

```python
performance_by_format_df = (
    chunking_benchmark_details_df
    .groupby(
        [
            "configuration",
            "format"
        ],
        as_index=False
    )
    .agg(
        number_of_questions=(
            "question_id",
            "count"
        ),

        evidence_recall_at_1=(
            "evidence_recall_at_1",
            "mean"
        ),

        evidence_recall_at_3=(
            "evidence_recall_at_3",
            "mean"
        ),

        evidence_recall_at_5=(
            "evidence_recall_at_5",
            "mean"
        ),

        evidence_recall_at_10=(
            "evidence_recall_at_10",
            "mean"
        ),

        evidence_mrr=(
            "evidence_reciprocal_rank",
            "mean"
        ),

        source_recall_at_5=(
            "source_recall_at_5",
            "mean"
        ),

        mean_retrieval_time_ms=(
            "retrieval_time_seconds",
            lambda values:
                values.mean() * 1000
        )
    )
)


display(
    performance_by_format_df.round(4)
)
```

**le tableau comparatif final**

```python
comparison_columns = [
    "configuration",
    "strategy",
    "parameter",
    "number_of_chunks",

    "evidence_recall_at_1",
    "evidence_recall_at_3",
    "evidence_recall_at_5",
    "evidence_recall_at_10",
    "evidence_mrr",

    "source_recall_at_1",
    "source_recall_at_3",
    "source_recall_at_5",
    "source_recall_at_10",
    "source_mrr",

    "mean_retrieval_time_ms",
    "p95_retrieval_time_ms",

    "embedding_time_seconds",
    "faiss_indexing_time_seconds",
    "total_configuration_time_seconds"
]


chunking_comparison_df = (
    chunking_benchmark_summary_df[
        comparison_columns
    ]
    .sort_values(
        by=[
            "evidence_recall_at_5",
            "evidence_mrr",
            "embedding_time_seconds"
        ],
        ascending=[
            False,
            False,
            True
        ]
    )
    .reset_index(
        drop=True
    )
)


display(
    chunking_comparison_df.round(4)
)
```
